# EcoPlots Python Library
# Use case 2 - Plant occurrence - vegetation height


```python
from terndata.ecoplots import EcoPlots
ecoplots = EcoPlots()

# Display the client information
print(ecoplots)
```

```python
ecoplots.select(feature_type="plant occurrence")
ecoplots.get_observed_properties()
```

```python

# Apply some filters
ecoplots.select(feature_type="plant occurrence").select(observed_property=["scientific name", "plant height", "growth form"])
# List available data sources
ecoplots.get_datasources()
```

```python
# Summarise the data before downloading it into a DataFrame
ecoplots.summary()
```

```python
ecoplots.select(site_id="NSTSYB0001")
```

```python
# Summarise the data before downloading it into a DataFrame
ecoplots.summary()
```

```python
# Download the data into a DataFrame. By default, this will return a pandas DataFrame.
# The total ammount of data downloaded will be 244k rows. This may take a few moments.
df_plant = ecoplots.get_data()
```

```python
# Display the first few rows of the DataFrame
df_plant.head()
```

```python
# Count occurrences of each scientific name, sorted in descending order (most frequent first)
df_plant.groupby(by=['plantGrowthForm']).size().sort_values(ascending=False)
```

```python
# Count occurrences of each scientific name, sorted in descending order (most frequent first)
df_plant.groupby(by=['scientificName']).size().sort_values(ascending=False)
```

```python
# transform plantHeight_metre to numeric, coerce errors to NaN
import pandas as pd
df_plant['plantHeight_metre'] = pd.to_numeric(df_plant['plantHeight_metre'], errors='coerce')
df_plant['plantHeight_metre'].dtype
```

```python
# Tallest plant individual for each scientific name
df_plant[['scientificName', 'plantHeight_metre']].loc[df_plant.groupby('scientificName')['plantHeight_metre'].idxmax()]
```

```python
# Group by growthForm and scientificName and calculate the average plant height for each growth form
df_plant.groupby(['plantGrowthForm'])['plantHeight_metre'].mean().reset_index()
```

```python
# Find the unique tallest plant from the dataframe
df_plant.loc[df_plant['plantHeight_metre'].idxmax()]
```

```python
# Find the tallest plant in each growth form
df_plant.loc[df_plant.groupby('plantGrowthForm')['plantHeight_metre'].idxmax()]
```
